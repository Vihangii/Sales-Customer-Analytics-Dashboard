```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:/Projects/Sales & Customer Analytics Dashboard/train.csv")

print("Shape:", df.shape)
print("\nDtypes:\n", df.dtypes)
print("\nNull counts:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Dates are in DD/MM/YYYY format — must parse explicitly or pandas
# will silently misread day/month for the first 12 days of each month
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

print("Order Date range:", df['Order Date'].min(), "to", df['Order Date'].max())
print("Ship Date range:", df['Ship Date'].min(), "to", df['Ship Date'].max())
```

```python
invalid_dates = df[df['Ship Date'] < df['Order Date']]
print("Rows where Ship Date is before Order Date:", len(invalid_dates))
invalid_dates
```

```python
dupes = df[df.duplicated(subset=['Order ID', 'Product ID'], keep=False)]
dupes = dupes.sort_values(['Order ID', 'Product ID'])
print("Rows involved in duplicate Order ID + Product ID:", len(dupes))
dupes
```

```python
print("Missing Postal Code rows:")
print(df[df['Postal Code'].isnull()][['City', 'State', 'Region']])

# Leaving as null — SQL Server can store this as NULL, and it's more honest
# than guessing a postal code. Documented as a known data limitation.
```

```python
df['Year'] = df['Order Date'].dt.year
df['Quarter'] = 'Q' + df['Order Date'].dt.quarter.astype(str)
df['Month'] = df['Order Date'].dt.strftime('%B')
df['Month Number'] = df['Order Date'].dt.month
df['Year-Month'] = df['Order Date'].dt.strftime('%Y-%b')

df[['Order Date', 'Year', 'Quarter', 'Month', 'Month Number', 'Year-Month']].head()
```

```python
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

print(df['Shipping Days'].describe())
df[['Order Date', 'Ship Date', 'Shipping Days']].head()
```

```python
print("Final shape:", df.shape)
print("Final columns:", list(df.columns))
print("\nRemaining nulls:\n", df.isnull().sum())

df.to_csv("data/cleaned/train_cleaned.csv", index=False)
print("\nSaved to data/cleaned/train_cleaned.csv")
```
